```python

import pandas as pd

date_time=["22-10-2026 16:21","21-10-2026 4:23","17-09-2025 10:30"]

df=pd.DataFrame({"Date_Time":date_time})

df
```

```python
#now  i want to create sepearate date / month/year / time 

#i want to fetch spearte separate use :-> to_datetime function:->  

#Permenat changes will be done after updating 

df['Date_Time']=pd.to_datetime(df['Date_Time'])

```

```python
#Create a year col 
df['Year']=pd.to_datetime(df['Date_Time']).dt.year #dt is inbuilt function

df
```

```python
#create a month col
df['Month']=pd.to_datetime(df['Date_Time']).dt.month
df
```

```python
#create a day
df['Day']=df['Date_Time'].dt.day
df
```

```python
#create a month name col
df['Month_name ']=df['Date_Time'].dt.month_name()
df
```

```python
#create a hour col
df['Hour']=df['Date_Time'].dt.hour
df
```

```python
#create a minute col
df["Minute"]=df['Date_Time'].dt.minute
df
```

```python
#create day name col
df["DayName"]=df['Date_Time'].dt.day_name()
df
```

```python
#create second col
df["Second"]=df['Date_Time'].dt.second
df
```

```python
#create weekday col
df['Week_Day']=df['Date_Time'].dt.day_of_week
df
```
