```python
'''1. Rank method:-> start from 1
        Syntax:-> varname.rank #s.rank

        -> smaller number 1st rank

        -> rank method by default mean

        -> in rank :-> method:-> 5 argument
        1. average
        2. dense :-> you have duplicated mens first ele postion will be display
        3. max
        4. min
        5. first
        
'''
```

```python
import pandas as pd

z=pd.Series([45,35,78,90,85,75])

print(z)
```

```python
z.rank()
```

```python
z.rank(ascending=False) #85,90,75,35,45,78
```

```python
x=pd.Series([78,90,85,65,45,23])
#            4, 6 ,5 ,3, 2, 1
x
```

```python
x.rank()
```

```python
x.rank(ascending=False) #65,23,45,78,85,90
```

```python
s=pd.Series([55,75,83,83,90,95])
s
```

```python
s.rank(method='average')
```

```python
s.rank(method='max')
```

```python
s.rank(method='min')
```

```python
s.rank(method='dense') #whenver we use it will give output as it is 
```

```python
s.rank(method='first')#whenver we applying first then first come first serve
```

```python


#using data frane

name=["Ravi","kiran","ram","naveen","krisha"]
marks=[50,70,89,90,90]
data={"NAME": name,"MARKS":marks}

df=pd.DataFrame(data)
df
```

```python
#
df["Rank"]=df["MARKS"].rank()
df
```

```python
df["MAX"]=df["MARKS"].rank(method='max')
df
```

```python
df["MIN"]=df["MARKS"].rank(method='min')
df
```

```python
df["DENSE"]=df["MARKS"].rank(method='dense')
df
```

```python
df["FIRST"]=df["MARKS"].rank(method='first')
df
```

```python
df["AVERAGE"]=df["MARKS"].rank(method='average')
df
```
