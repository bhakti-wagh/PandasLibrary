```python
'''Merge:-> if we want to combine two table
    Syntax:->   newvarName=pd.merge(leftTable,rightTable,ON=column,how=Type of join)-> by default inner join

    -> 
    
    '''
```

```python
import pandas as pd

name=["A","B","C","D","E"]
id=[1,2,3,4,5]

tabel1={"NAME":name,"ID":id}

df1=pd.DataFrame(tabel1)
df1
```

```python
df2=pd.DataFrame({"ID":[2,3,4,6,7],"MARKS":[45,90,66,35,75]})
df2
```

```python
res=pd.merge(df1,df2,on="ID")
res
```

```python
res1=pd.merge(df1,df2,on="ID", how="inner")
res1
```

```python
res2=pd.merge(df1,df2,on="ID", how="right")
res2
```

```python
res3=pd.merge(df1,df2,on="ID", how="left")
res3
```

```python
res4=pd.merge(df1,df2,on="ID",how="outer")
res4
```

```python
t1=pd.DataFrame({"ID":[2,3,4,6,7], "name": ["A", "B", "C", "D", "E"]})
t2 = pd.DataFrame({"MARKS":[45,90,66,35,75],"Rank": ["*", 2, 3, 4, 5]})


```

```python
result=pd.merge(t1,t2,left_on="ID",right_on="MARKS",how="outer")
result
```

```python
res=pd.concat([t1,t2],axis=1) #1-> col ,0-> row
res
```
