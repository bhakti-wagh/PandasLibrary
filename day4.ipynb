```python
import pandas as pd

data={
    "name":["A","B","C","D"],
    "age":[23,45,66,78],
    "sal":[1000,2000,3000,4000],
    "department":["HR","Analysis","HR","IT"]
}
df=pd.DataFrame(data)
df
```

```python
df.groupby("department")
```

```python
df.groupby("department")["sal"].value_counts()
```

```python
df.groupby("department")["sal"].sum()
```

```python
df.groupby("department")["sal"].mean()
```

```python
df.groupby("department")["sal"].count()
```

```python
df.groupby("department")["sal"].agg(["min","sum","max","count"])
```
